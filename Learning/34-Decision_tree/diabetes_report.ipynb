## diabetes reports regression

```python
import numpy as np
import pandas as pd 
```

```python
from sklearn.datasets import load_diabetes
data= load_diabetes()
```

```python
data
```

```python
print(data['DESCR'])
```

```python
data
```

```python
data['target']
```

```python
X=pd.DataFrame(data.data, columns=['age', 'sex', 'bmi', 'bp','s1', 's2', 's3', 's4', 's5', 's6'])
y = data['target']
```

```python
pd.DataFrame(X).head()
```

```python
from sklearn.model_selection import train_test_split 
X_train , X_test, Y_train , Y_test= train_test_split(X, y, train_size= 0.25, random_state= 10)
```

```python
X_train.head()
```

```python
X_train.corr()
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 10))
sns.heatmap(X_train.corr(), annot=True)
```

```python
from sklearn.tree import DecisionTreeRegressor
tree= DecisionTreeRegressor()
tree.fit(X_train, Y_train)      
```

```python
param={
    'criterion': ['absolute_error', 'squared_error', 'poisson'],
    'splitter': ['best', 'random'],
    'max_depth' : [1,2,3,4,5],
    'max_features':['auto', 'sqrt','log2']
 }
```

```python
from sklearn.model_selection import GridSearchCV
model= GridSearchCV(estimator=tree , param_grid= param, cv= 5,scoring='neg_mean_squared_error' )
```

```python
model.fit(X_train, Y_train)
```

```python
y_pred=model.predict(X_test)
print(y_pred)
```

```python
model.best_params_
```

```python
model.best_score_
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
print(r2_score(Y_test, y_pred))
print(mean_absolute_error(Y_test, y_pred))
print(mean_squared_error(Y_test, y_pred))
```

```python
from sklearn.tree import DecisionTreeRegressor
regressorTree= DecisionTreeRegressor(criterion='squared_error', max_depth=3, max_features='sqrt', splitter='random')
```

```python
regressorTree.fit(X_train, Y_train)
```

```python
## visualize the decision tree
import matplotlib.pyplot as plt
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(regressorTree, filled= True)
```
