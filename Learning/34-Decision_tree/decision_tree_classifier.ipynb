### Decision tree classifier 

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline 
```

```python
from sklearn.datasets import load_iris
```

```python
iris= load_iris()
```

```python
print(iris)
```

```python
print(iris['DESCR'])
```

```python
print(iris['target'])
```

```python
##independent feature 
X=pd.DataFrame(iris['data'],columns=['sepal length in cm', 'sepal width in cm','petal length in cm','petal width in cm'])
```

```python
y= iris['target']
```

```python
from sklearn.model_selection import train_test_split
X_train ,X_test, y_train , y_test = train_test_split(X,y, random_state=42, test_size=0.25)
```

```python
## apply decision tree classifier
from sklearn.tree import DecisionTreeClassifier
treeClassifier= DecisionTreeClassifier(max_depth= 3)
```

```python
treeClassifier.fit(X_train, y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(treeClassifier, filled=True)
```

```python
y_predict= treeClassifier.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print(accuracy_score(y_test, y_predict))
print(confusion_matrix(y_test, y_predict))
print(classification_report(y_test, y_predict))
```

## Decision tree classifier preprunning and hyper parameter tunning


```python
param={
    'criterion': ['gini', 'entropy','log_loss'],
    'splitter': ['best', 'random'],
    'max_depth' : [1,2,3,4,5],
    'max_features':['auto', 'sqrt','log2']
 }
```

```python
from sklearn.model_selection import GridSearchCV
treemodel = DecisionTreeClassifier()
grid= GridSearchCV(treemodel,param_grid= param,cv=5, scoring='adjusted_rand_score')
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred=grid.predict(X_test)

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print(accuracy_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```
