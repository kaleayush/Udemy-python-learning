```python
## independent and dependent features 
X= df[['tip', 'sex', 'smoker', 'day', 'time', 'size']]
Y= df['total_bill']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test= train_test_split(X,Y, test_size=0.25,random_state=0)
```

```python
X_train.head()
```

```python
## feature encoding(label encoding onehot encoding)
from sklearn.preprocessing import LabelEncoder
```

```python
le1 = LabelEncoder()
le2 = LabelEncoder()
le3 = LabelEncoder()
```

```python
# import warnings
# warnings.filterwarnings('ignore')
X_train['sex']=le1.fit_transform(X_train['sex'])
X_train['smoker']=le1.fit_transform(X_train['smoker'])
X_train['time']=le1.fit_transform(X_train['time'])
```

```python
X_train.head()
```

```python
X_test['sex']=le1.transform(X_test['sex'])
X_test['smoker']=le1.transform(X_test['smoker'])
X_test['time']=le1.transform(X_test['time'])
```

```python
X_test.head()
```

```python
## onehot encoding
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct=ColumnTransformer(transformers=[('onehot',OneHotEncoder(drop='first'),[3] )],remainder='passthrough')
```

```python
import sys
import numpy
numpy.set_printoptions(threshold=sys.maxsize)
X_train=ct.fit_transform(X_train)

```

```python
X_train
```

```python
X_test=ct.transform(X_test)
```

```python
X_test
```

```python
from sklearn.svm import SVR
svr=SVR()
```

```python
svr.fit(X_train,Y_train)
```

```python
y_pred=svr.predict(X_test)
```
