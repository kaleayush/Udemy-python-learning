```python
import pandas as pd 
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification
X,Y = make_classification(n_samples=1000, n_features=2, n_classes= 2, n_clusters_per_class=1, n_redundant=0)
X
```

```python
pd.DataFrame(X)[0]
```

```python
Y
```

```python
df = pd.DataFrame(X, columns=['Feature1', 'Feature2'])
print(df)
sns.scatterplot(data=df, x='Feature1', y='Feature2', hue=Y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train , Y_test= train_test_split(X, Y, test_size=0.25, random_state=10)
```

```python
from sklearn.svm import SVC
```

```python
svc= SVC(kernel='linear')
```

```python
svc.fit(X_train, Y_train)
```

```python
Y_predit=svc.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(Y_test, Y_predit))
print(confusion_matrix(Y_test, Y_predit))
```

```python
# SVC with RBF kernal
svc= SVC(kernel='rbf')
svc.fit(X_train, Y_train)
Y_predit1=svc.predict(X_test)
print(classification_report(Y_test, Y_predit1))
print(confusion_matrix(Y_test, Y_predit1))
```

```python
# SVC with poly kernal
svc= SVC(kernel='poly')
svc.fit(X_train, Y_train)
Y_predit1=svc.predict(X_test)
print(classification_report(Y_test, Y_predit1))
print(confusion_matrix(Y_test, Y_predit1))
```

```python
# SVC with RBF kernal
svc= SVC(kernel='sigmoid')
svc.fit(X_train, Y_train)
Y_predit1=svc.predict(X_test)
print(classification_report(Y_test, Y_predit1))
print(confusion_matrix(Y_test, Y_predit1))
```

### Hyper parameter tunning in svc

```python
from sklearn.model_selection import GridSearchCV
```

```python
param= {'C': [0.1,1,10,100,1000], 'gamma' : [1, 0.1, 0.01, 0.001, 0.0001], 'kernel': ['rbf']}
```

```python
grid= GridSearchCV(SVC(),param_grid= param, refit=True, cv=5, verbose=3)
```

```python
grid.fit(X_train, Y_train)
```

```python
grid.best_score_
```

```python
grid.best_params_
```

```python
Y_predit1=grid.predict(X_test)
print(classification_report(Y_test, Y_predit1))
print(confusion_matrix(Y_test, Y_predit1))
```
