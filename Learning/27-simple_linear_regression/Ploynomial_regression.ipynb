```python
##import library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
X= 6 * np.random.rand(100,1) - 3
Y = 0.5 * X**2 + 1.5 * X + 2 + np.random.rand(100,1)
## this is the quadratic equation y  = ax**2 + bx + c
```

```python
plt.scatter(X, Y ,color= 'r')
plt.xlabel('x dataset')
plt.ylabel('y dataset')
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test= train_test_split(X,Y ,test_size=0.25, random_state= 42)
```

```python
# lets implemnet simple linear regression
from sklearn.linear_model import LinearRegression
```

```python
X_train
```

```python
regression = LinearRegression()
```

```python
regression.fit(X_train, Y_train)
```

```python
from sklearn.metrics import r2_score
```

```python
score = r2_score(Y_test, regression.predict(X_test))
```

```python
print(score)
```

```python
##lets visualize this model
plt.plot(X_train, regression.predict(X_train), color='r')
plt.scatter(X_train, Y_train)
plt.xlabel('X dataset')
plt.ylabel('y dataset')
```

```python
## lets apply polynimial regression
from sklearn.preprocessing import PolynomialFeatures
```

```python
ploy= PolynomialFeatures(degree= 2 , include_bias= True)
X_ploy_train= ploy.fit_transform(X_train)
X_ploy_test= ploy.transform(X_test)
```

```python
X_ploy_train
```

```python
from sklearn.metrics import r2_score
regression= LinearRegression()
regression.fit(X_ploy_train, Y_train)
Y_predict=regression.predict(X_ploy_test)
score= r2_score(Y_test, Y_predict)
print(score)
```

```python
print(regression.coef_)
```

```python
print(regression.intercept_)
```

```python
plt.scatter(X_train, regression.predict(X_ploy_train))
plt.scatter(X_train, Y_train)
```

```python
## prediction for new data set
X_new = np.linspace(-3, 3, 200).reshape(200,1)
X_new_ploy= ploy.transform(X_new)
X_new_ploy
```

```python
Y_new = regression.predict(X_new_ploy)
plt.plot(X_new, Y_new, "r-", linewidth= 2 , label= "Predictions")
plt.plot(X_train, Y_train, "b.", label='Training points')
plt.plot(X_test, Y_test, 'g.', label= "test points")
plt.xlabel("x")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
from sklearn.pipeline import Pipeline
```

```python
def ploy_regression(degree):
    X_new = np.linspace(-3, 3 ,200).reshape(200,1)

    ploy_feature= PolynomialFeatures(degree= degree, include_bias= True)

    lin_reg= LinearRegression()
    poly_regression = Pipeline([
        ("ploy_features", ploy_feature),
        ("lin_reg", lin_reg )
        ])
    poly_regression.fit(X_train, Y_train)
    Y_predict= poly_regression.predict(X_new)
   #plotting prediction line
    plt.plot(X_new, Y_predict,'r', label="Degree " + str(degree), linewidth=2)
    plt.plot(X_train, Y_train, "b.", linewidth=3)
    plt.plot(X_test, Y_test, "g.", linewidth=3)
    plt.legend(loc="upper left")
    plt.xlabel("X")
    plt.ylabel("y")
    plt.axis([-4,4, 0, 10])
    plt.show()
```

```python
ploy_regression(15)
```
