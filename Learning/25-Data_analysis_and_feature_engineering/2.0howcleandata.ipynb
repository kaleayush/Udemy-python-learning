```python
import pandas as pd
import numpy as np  
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
##feature engineering

df['Date'] = df['Date_of_Journey'].str.split('/').str[0]
df['Month'] = df['Date_of_Journey'].str.split('/').str[1]
df['Year'] = df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.head()
df.info()
```

```python
df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df['Year']=df['Year'].astype(int)
```

```python
df.info()
```

```python
#drop date of journey
df.drop('Date_of_Journey', axis=1, inplace=True)
```

```python
df.head()
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
```

```python
df['Arrival_hour']= df['Arrival_Time'].str.split(":").str[0]
df['Arrival_min']= df['Arrival_Time'].str.split(":").str[1]
```

```python
df.head(2)
```

```python
df.info()
```

```python
df['Arrival_hour']=df['Arrival_hour'].astype(int)
df['Arrival_min']=df['Arrival_min'].astype(int)
```

```python
df.drop('Arrival_Time', axis=1 , inplace=True)
```

```python
df.info()
```

```python
df['Departure_hour']= df['Dep_Time'].str.split(":").str[0]
df['Departure_min']= df['Dep_Time'].str.split(":").str[1]
```

```python
df.head()
```

```python
df['Departure_hour']= df['Departure_hour'].astype(int)
df['Departure_min']= df['Departure_min'].astype(int)
```

```python
df.head()
```

```python
df.drop('Dep_Time', axis=1, inplace=True)
```

```python
df.head()
```

```python
df.info()
```

```python
df['Total_Stops'].unique()
```

```python
df[df["Total_Stops"].isnull()]
```

```python
df['Total_Stops']=df['Total_Stops'].map({'non-stop': 0, '2 stops':2, '1 stop' :1, '3 stops' : 3, '4 stops':4,np.nan:1})
```

```python
df[df["Total_Stops"].isnull()]
```

```python
df.head(2)
```

```python
df.drop('Route', axis=1, inplace=True)
```

```python
df.info()
```

```python
df.head(2)
```

```python
df['Duration']= df['Duration'].str.replace('h','')
df['Duration']= df['Duration'].str.replace('m','')
```

```python
df['Duration']
```

```python
df['Duration_hour']= df['Duration'].str.split(' ').str[0]
df['Duration_min']= df['Duration'].str.split(' ').str[1]
```

```python
df['Duration_min']= df['Duration_min'].replace(np.nan,0)
```

```python
df.drop('Duration', axis=1, inplace=True)
```

```python
df['Duration_hour'] = df['Duration_hour'].astype(int)
df['Duration_min'] = df['Duration_min'].astype(int)
df.info()
```
